# Jupyter Notebook Practice: Loading and Inspecting a Dataset

This notebook is part of issue #20 (Using Jupyter Notebooks for Interactive Analysis).
The goal here is just to practice the core workflow: load a dataset, inspect it,
and document findings using a mix of code cells and markdown cells.

## Step 1: Load the dataset

For this practice run I'm using a small sample dataset built directly in the notebook,
so it's self contained and doesn't depend on an external file.

```
import pandas as pd

# Small sample dataset representing daily focus session data,
# similar in shape to what a real usage dataset might look like
data = {
    "user_id": [1, 2, 3, 4, 5, 6],
    "date": pd.date_range("2026-07-01", periods=6, freq="D"),
    "focus_minutes": [45, 60, 30, 90, None, 75],
    "sessions_completed": [2, 3, 1, 4, 2, 3],
    "device": ["desktop", "mobile", "desktop", "mobile", "desktop", "mobile"],
}

df = pd.DataFrame(data)
df
```

## Step 2: Inspect the dataset

Basic checks before doing anything else: shape, column types, and whether
there's any missing data to be aware of.

```
print("Shape:", df.shape)
print()
print(df.dtypes)
print()
print("Missing values per column:")
print(df.isnull().sum())
```

## Step 3: A quick summary

Getting basic descriptive stats on the numeric columns.

```
df.describe()
```

## Insights

- There's one missing value in `focus_minutes` (user 5), which would need to be
  handled (e.g. filled with the mean, or excluded) before any further analysis.
- Average focus time and session counts differ a bit by device in this small
  sample, worth checking if that pattern holds with a larger, real dataset.
- This kind of quick load -> inspect -> summarise loop is exactly the workflow
  Jupyter is good for: each step runs in its own cell, and I can see the output
  immediately without re-running the whole script.
